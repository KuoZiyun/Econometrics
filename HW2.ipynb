```python
#資料抓取
import requests
from bs4 import BeautifulSoup as bs
import pandas as pd
import datetime as dt
import time #超過16筆資料，休息10秒

def countdown(t):
  while t > -1:
    min, sec = divmod(t, 60)
    if t > 0:
      print("{:02}:{:02}".format(min, sec), end = "\r")
    elif t == 0:
      print("{:02}:{:02}".format(min, sec), end = "\n")
    t = t - 1
    time.sleep(1)

url = "https://www.twse.com.tw/rwd/zh/fund/BFI82U"
payload = {
    "response":"html",
    "dayDate":"20251008"
}

data = []
for i in range(16):
  day = dt.datetime.today() - dt.timedelta(days = i + 1) #查詢變更日期
  print(day.strftime("%Y%m%d"))

  payload["dayDate"] = day.strftime("%Y%m%d") #修改參數
  res = requests.get(url, params = payload) #連線抓取資料
  if res.text.find("三大法人買賣金額統計表") > -1: #判斷是否有資料，防呆
    soup = bs(res.text, "lxml") #解析網頁原始程式碼
    tb = soup.select("table")[0] #抓取網頁表格
    df0 = pd.read_html(str(tb), header = [1])[0].iloc[:4] #解析網頁表格，存入DataFrame
    df0.insert(0, "日期", day.strftime("%Y%m%d"))
    data.append(df0) #抓取表格
  #countdown(10) #休息10秒
df = pd.concat(data) #合併所有表格
df

#Google colab安裝中文字型
!wget -O TaipeiSansTCBeta-Regular.ttf https://drive.google.com/uc?id=1eGAsTN1HBpJAkeVM57_C7ccp7hbgSz3_&export=download

import matplotlib
matplotlib.font_manager.fontManager.addfont('TaipeiSansTCBeta-Regular.ttf')
matplotlib.rc('font', family='Taipei Sans TC Beta')

#製作圖表
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams["font.family"] = ['Taipei Sans TC Beta']

df['買賣差額2'] = df["買賣差額"]/1000000 #轉換數值
自營商 = df[df["單位名稱"] == "自營商(自行買賣)"]
投信 = df[df["單位名稱"] == "投信"]
外資 = df[df["單位名稱"] == "外資及陸資(不含外資自營商)"]
w = 0.25
plt.figure(figsize = (11, 5))
plt.bar(np.arange(len(自營商)), height = 自營商["買賣差額2"], width = w)
plt.bar(np.arange(len(自營商)) + w, height = 投信["買賣差額2"], width = w)
plt.bar(np.arange(len(自營商)) + 2*w, height = 外資["買賣差額2"], width = w)
plt.xticks(np.arange(len(自營商))*1.07, 自營商["日期"], rotation = 30) #設定x軸刻度  <- 決定X軸間隔
plt.legend(["自營商", "投信", "外資"]) #圖例設定
plt.title("三大法人買賣金額統計表", size = 18)
plt.xlabel("日期", size = 14)
plt.ylabel("金額(單位:百萬)", size = 14)
plt.savefig("三大法人買賣金額統計表.png", format="png", dpi = 200, bbox_inches = "tight")
```
