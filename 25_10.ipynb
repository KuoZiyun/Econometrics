```python
import requests
from bs4 import BeautifulSoup as bs
import pandas as pd
import datetime as dt
import time #超過16筆資料，休息10秒

def countdown(t):
  while t > -1:
    min, sec = divmod(t, 60)
    if t > 0:
      print("{:02}:{:02}".format(min, sec), end = "\r")
    elif t == 0:
      print("{:02}:{:02}".format(min, sec), end = "\n")
    t = t - 1
    time.sleep(1)

url = "https://www.twse.com.tw/rwd/zh/fund/BFI82U"
payload = {
    "response":"html",
    "dayDate":"20251008"
}

data = []
for i in range(5):
  day = dt.datetime.today() - dt.timedelta(days = i + 1) #查詢變更日期
  print(day.strftime("%Y%m%d"))

  payload["dayDate"] = day.strftime("%Y%m%d") #修改參數
  res = requests.get(url, params = payload) #連線抓取資料
  if res.text.find("三大法人買賣金額統計表") > -1: #判斷是否有資料，防呆
    soup = bs(res.text, "lxml") #解析網頁原始程式碼
    tb = soup.select("table")[0] #抓取網頁表格
    df0 = pd.read_html(str(tb), header = [1])[0].iloc[:4] #解析網頁表格，存入DataFrame
    df0.insert(0, "日期", day.strftime("%Y%m%d"))
    data.append(df0) #抓取表格
  #countdown(10) #休息10秒
df = pd.concat(data) #合併所有表格
df
```

```python
from datetime import date
from calendar import monthrange

def generate_monthly_date_list():
    today = date.today()
    one_year_ago = date(today.year - 1, today.month, 1)  # 往前一年、當月第一天

    result = []
    year = one_year_ago.year
    month = one_year_ago.month

    while (year < today.year) or (year == today.year and month <= today.month):
        first_day = date(year, month, 1)
        last_day = date(year, month, monthrange(year, month)[1])

        # 格式化成 "YYYY/MM/DD"
        start_str = first_day.strftime("%Y/%m/%d")
        end_str = last_day.strftime("%Y/%m/%d")

        result.append((start_str, end_str))

        # 移動到下一個月
        if month == 12:
            month = 1
            year += 1
        else:
            month += 1

    return result


# 示範輸出
if __name__ == "__main__":
    date_list = generate_monthly_date_list()
    for start, end in date_list:
        print(f"{start} ~ {end}")
```

```python
import datetime as dt
from dateutil.relativedelta import relativedelta

base = dt.datetime.today() - relativedelta(years = 1) #日期時間計算
while base < dt.datetime.today(): #日期時間比較
  startdate = base.replace(day = 1).strftime("%Y/%m/%d") #日期時間取代與格式轉換
  enddate = (base.replace(day = 1) + relativedelta(months = 1) - relativedelta(days = 1)).strftime("%Y/%m/%d")
  print(startdate + " ~ " + enddate)
  base = base + relativedelta(months = 1) #月份累加
```

```python
import datetime as dt
from dateutil.relativedelta import relativedelta
import requests
from bs4 import BeautifulSoup as bs
import pandas as pd
from io import StringIO

def pcratio(startdate, enddate):
  url = "https://www.taifex.com.tw/cht/3/pcRatio"
  payload = {
    "down_type":"",
    "queryStartDate":startdate,
    "queryEndDate":enddate
  }
  res = requests.get(url, params = payload)
  soup = bs(res.text, "lxml")
  tb = soup.select("table")[0]
  df = pd.read_html(StringIO(tb.prettify()))[0]
  return df

data = []
base = dt.datetime.today() - relativedelta(years = 1) #日期時間計算
while base < dt.datetime.today(): #日期時間比較
  startdate = base.replace(day = 1).strftime("%Y/%m/%d") #日期時間取代與格式轉換
  enddate = (base.replace(day = 1) + relativedelta(months = 1) - relativedelta(days = 1)).strftime("%Y/%m/%d")
  print(startdate + " ~ " + enddate)
  data.append(pcratio(startdate, enddate))
  base = base + relativedelta(months = 1) #月份累加
df = pd.concat(data, ignore_index = True)
df
```
